```python
# PRACTICE — NOT REFERENCE DATA
"""Day 1 Practice: Pure Python implementation of Curatr scoring formulas.
Verifies the 5-movie fictional hand-check dataset from Section 5.
"""

def global_score(rating: float, vote_count: int, C: float, m: float) -> float:
    """Bayesian weighted score pulling raw rating toward global prior C."""
    return (vote_count * rating + m * C) / (vote_count + m)

def genre_score(rating: float, vote_count: int, C: float, m: float, G: float) -> float:
    """Genre-adjusted score evaluating performance relative to genre baseline G."""
    return C + (vote_count / (vote_count + m)) * (rating - G)

# Fictional 5-movie parameters from Section 5
C = 7.2000
m = 40.0000

# Fictional movies: (title, R, v, G)
movies = [
    ("Star Horizon", 8.4, 200, 7.0167),
    ("Fast Track", 6.2, 40, 7.3333),
    ("Solitary Peak", 9.0, 10, 8.2000),
    ("Neon Void", 5.0, 50, 6.7000),
    ("Iron Heart", 7.4, 100, 7.7667),
]

print(f"{'Title':<15} {'R':<5} {'v':<5} {'S_global':<10} {'S_genre':<10}")
print("-" * 50)
for title, r, v, g in movies:
    s_glob = global_score(r, v, C, m)
    s_gen = genre_score(r, v, C, m, g)
    print(f"{title:<15} {r:<5.1f} {v:<5} {s_glob:<10.4f} {s_gen:<10.4f}")
# Hand-check assertion for Star Horizon (Section 5 master table)
s_glob_star = global_score(8.4, 200, C, m)
s_gen_star = genre_score(8.4, 200, C, m, 7.0167)
assert abs(s_glob_star - 8.2000) < 1e-4, f"Expected 8.2000, got {s_glob_star}"
assert abs(s_gen_star - 8.3528) < 1e-4, f"Expected 8.3528, got {s_gen_star}"
print("\nHand-check assertions passed!")
```
